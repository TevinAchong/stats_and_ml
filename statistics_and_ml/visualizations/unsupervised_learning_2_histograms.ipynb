# Goal
Convert the original histogram from count to proportion.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# number of data points
n = 1000

# generate data - log-normal distribution
data = np.exp(np.random.randn(n) / 2)
```

```python
# Plot histogram counts 
k = 40
plt.hist(data, bins=k)
plt.show()
```

```python
np.histogram(data, bins=k)
```

The first array gives the height of each of the bars. 

So we need to convert these heights to proportions I think.

```python
counts = np.histogram(data, bins=k)[0]

print(f"Total Number of Heights: {len(counts)}")
print(f"Sum of all heights: {counts.sum()}")
```

```python
proportions = []
for i in counts:
    proportions.append((i / counts.sum())*100)
proportions = np.array(proportions)
proportions
```

```python
proportions.sum()
```

```python
data_bins = np.histogram(data, bins=k)[1]
centers = (data_bins[1:]+data_bins[:-1])/2
centers
```

```python
plt.plot(centers, proportions)
plt.show()
```
