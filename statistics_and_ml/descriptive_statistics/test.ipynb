## Goal - Produce an Asymmetric Violin Plot

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
from scipy import stats
```

```python
# Lets just create a normally distributed dataset
normal = np.random.randn(50)
normal
```

```python
plt.violinplot(normal)
```

```python
normal_two = np.random.randn(50)
normal_two
```

```python
plt.violinplot([normal, normal_two])
plt.xticks([1, 2], ["Normal One", "Normal Two"])
```

```python
import seaborn as sns
import pandas as pd
```

```python
normal_df = pd.DataFrame ({
    "value" : np.r_[normal, normal_two],
    "group" : ["One"] * len(normal) + ["Two"] * len(normal_two),
    "x" : ["same"] * (len(normal) + len(normal_two))
})
normal_df.head()
```

```python
# Explaining np.r_
arr1 = np.array([1,2,3])
arr2 = np.array([4, 5, 6])
np.r_[arr1, arr2]
```

```python
sns.violinplot(data=normal_df, x="x", y="value", hue="group", split=True)
```
